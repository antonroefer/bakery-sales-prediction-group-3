```python
import pandas as pd

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)


# Sales data
umsatzdaten = pd.read_csv("../0_DataPreparation/Internal/umsatzdaten_gekuerzt.csv")

# Weather data
wetter = pd.read_csv("../0_DataPreparation/Internal/wetter.csv")

# Kieler Woche data
kiwo = pd.read_csv("../0_DataPreparation/Internal/kiwo.csv")

# Ensure date format is consistent
umsatzdaten["Datum"] = pd.to_datetime(umsatzdaten["Datum"])
kiwo["Datum"] = pd.to_datetime(kiwo["Datum"])
wetter["Datum"] = pd.to_datetime(wetter["Datum"])

# Merge the DataFrames on the 'Datum' column
merged_df = umsatzdaten.merge(kiwo, on="Datum", how="outer").merge(
    wetter, on="Datum", how="outer"
)

# Ensure 'Datum' to datetime and extract the year, day of the year, and weekday
merged_df["Datum"] = pd.to_datetime(merged_df["Datum"])

# Create a DataFrame with all combinations of Datum and Warengruppe (1-6)
all_dates = pd.DataFrame({"Datum": merged_df["Datum"].unique()})
all_warengruppen = pd.DataFrame({"Warengruppe": range(1, 7)})
all_combinations = all_dates.merge(all_warengruppen, how="cross")

# Merge with the original dataframe to ensure every date has all 6 Warengruppen
merged_df = all_combinations.merge(merged_df, on=["Datum", "Warengruppe"], how="left")

# Optional: sort by date and Warengruppe for readability
merged_df = merged_df.sort_values(["Datum", "Warengruppe"]).reset_index(drop=True)

# Constructing new variables for later use
merged_df["Jahr"] = merged_df["Datum"].dt.year
merged_df["Monat"] = merged_df["Datum"].dt.month
merged_df["Tag_im_Jahr"] = merged_df["Datum"].dt.dayofyear
merged_df["Wochentag"] = merged_df["Datum"].dt.weekday + 1  # 1=Monday, 7=Sunday

# in 'KielerWoche' fill NaN values  with 0
merged_df["KielerWoche"] = merged_df["KielerWoche"].fillna(
    0
)  # 0 = no Kieler Woche, 1 = Kieler Woche

# Set the ID in the format yymmddX (e.g., 1307053 for 2013-07-05, Warengruppe 3)
merged_df["id"] = merged_df["Datum"].dt.strftime("%y%m%d") + merged_df[
    "Warengruppe"
].astype(str)
merged_df["id"] = merged_df["id"].astype(int)


# Just for review ------------------------------------------------------------------


# Checking the data types
print("\nData types:")
print(merged_df.dtypes)

print(merged_df["id"])

merged_df.to_html("merged_df.html", index=False)
```

Here we created averages describing the avg temperature each month and then split over the years

```python
temp = wetter["Temperatur"].value_counts().sort_index()
print(len(temp))
plt.figure(figsize=(30, 6))
temp.plot(kind="line", marker="o")  # oder 'bar' für Balken
plt.title("Häufigkeit der Temperaturwerte")
plt.xlabel("Temperatur")
plt.ylabel("Anzahl")
plt.grid(True)
plt.show()
```

```python
cum_counts = temp.cumsum()

# Gesamtanzahl an Datenpunkten
total = cum_counts.iloc[-1]

# Anzahl gewünschter Gruppen (z. B. n = 3)
for n in range(2, 10):

# Schwellenwerte berechnen (1/n, 2/n ... der Gesamtdatenmenge)
    thresholds = [total * i / n for i in range(1, n)]

    # Temperaturwerte finden, bei denen die kumulierten Summen diese Schwellen überschreiten
    split_points = [cum_counts[cum_counts >= t].index[0] for t in thresholds]

    # Plot
    plt.figure(figsize=(30, 6))
    temp.plot(kind="line", marker="o")
    plt.title(f"Häufigkeit der Temperaturen mit {n} gleich großen Gruppen")
    plt.xlabel("Temperatur")
    plt.ylabel("Anzahl")
    plt.grid(True)

    # Trennlinien einzeichnen
    for x in split_points:
        plt.axvline(x=x, color="red", linestyle="--", label=f"Trennlinie bei {x}°C")

    plt.legend()
    plt.show()
```

```python
wetter_dict = {}

for i in range(1, 13):
    wetter_dict[i] = wetter[wetter["Monat"] == i]

for i in range(1, 13):  # Für jeden Monat
    # Häufigkeitstabelle der Temperatur im Monat i
    temp = wetter_dict[i]["Temperatur"].value_counts().sort_index()
    print(len(temp))
    # Überspringen, wenn keine Daten vorhanden sind
    if temp.empty:
        continue

    # Kumulative Häufigkeit berechnen
    cum_counts = temp.cumsum()
    total = cum_counts.iloc[-1]

    for n in range(2, 7):  # Gruppenanzahl 2 bis 10
        thresholds = [total * j / n for j in range(1, n)]
        split_points = [cum_counts[cum_counts >= t].index[0] for t in thresholds]

        # Plot
        plt.figure(figsize=(10, 5))
        temp.plot(kind="line", marker="o")
        plt.title(f"Monat {i}: Temperaturverteilung in {n} gleich großen Gruppen")
        plt.xlabel("Temperatur")
        plt.ylabel("Anzahl")
        plt.grid(True)

        for x in split_points:
            plt.axvline(x=x, color="red", linestyle="--", label=f"Trennlinie bei {x}°C")

        plt.legend(loc="upper right")
        plt.tight_layout()
        plt.show()
```
